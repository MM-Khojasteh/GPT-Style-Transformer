# Basic GPT-Style Transformer in Python

This notebook implements a basic GPT-style transformer model with token embeddings, positional embeddings, multi-head attention, and transformer blocks.


```python
import torch
import torch.nn as nn
from src.models.gpt import AdvancedGPT

# Create a basic GPT model
model = AdvancedGPT(
    vocab_size=1000,
    max_seq_len=128,
    embed_dim=128,
    num_heads=4,
    num_layers=2,
    dropout_prob=0.1,
    pos_emb_type='learned',
    use_rope=False
)

print(f"Model created with {model.get_num_params():,} parameters")

# Test forward pass
input_ids = torch.randint(0, 1000, (2, 64))
logits = model(input_ids)
print(f"Input shape: {input_ids.shape}")
print(f"Output logits shape: {logits.shape}")
```

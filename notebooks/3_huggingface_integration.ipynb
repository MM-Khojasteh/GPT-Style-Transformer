# Hugging Face Integration

This notebook demonstrates integration with Hugging Face datasets and tokenizers.


```python
from datasets import load_dataset
from transformers import AutoTokenizer
from src.models.gpt import AdvancedGPT

# Load dataset from Hugging Face
dataset = load_dataset("wikitext", "wikitext-2-raw-v1", split="train")
print(f"Loaded {len(dataset)} examples")

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained("gpt2")
print(f"Tokenizer vocab size: {tokenizer.vocab_size}")

# Create model matching tokenizer
model = AdvancedGPT(
    vocab_size=tokenizer.vocab_size,
    max_seq_len=1024,
    embed_dim=768,
    num_heads=12,
    num_layers=12,
    use_rope=True,
    weight_tying=True
)

print("Model ready for training with Hugging Face data!")
```

# Bigram Language Model in Python

This notebook implements a simple bigram language model that learns word pair frequencies and generates text based on those probabilities.


```python
import random
from collections import defaultdict
from typing import Dict, List


def build_bigram_model(corpus: str) -> Dict[str, Dict[str, int]]:
    """Build bigram model from corpus."""
    tokens = corpus.lower().split()
    model = defaultdict(lambda: defaultdict(int))
    
    # Count bigrams
    for i in range(len(tokens) - 1):
        current = tokens[i]
        next_word = tokens[i + 1]
        model[current][next_word] += 1
    
    return model


def generate_text(
    model: Dict[str, Dict[str, int]],
    start_word: str,
    num_words: int
) -> str:
    """Generate text from bigram model."""
    current_word = start_word
    output = [current_word]
    
    for _ in range(num_words - 1):
        if current_word not in model:
            break
        
        # Get next word options and their counts
        next_options = model[current_word]
        words = list(next_options.keys())
        counts = list(next_options.values())
        
        # Compute probabilities
        total = sum(counts)
        probs = [c / total for c in counts]
        
        # Sample next word
        current_word = random.choices(words, weights=probs)[0]
        output.append(current_word)
    
    return " ".join(output)


def predict_next_words(model: Dict[str, Dict[str, int]], word: str):
    """Predict next words with probabilities."""
    if word not in model:
        print(f"No predictions available for '{word}'.")
        return
    
    next_options = model[word]
    total = sum(next_options.values())
    
    print(f"Predicted next words for '{word}':")
    for w, count in sorted(next_options.items(), key=lambda x: x[1], reverse=True):
        prob = round(count / total, 2)
        print(f"  {w}: {prob}")
```

```python
# Example usage
corpus = """
This is a sample text for our bigram language model. 
This language model will help generate text based on bigram frequencies.
"""

model = build_bigram_model(corpus)
print("Generated text:")
print(generate_text(model, "this", 20))
```

```python
# Extended example
corpus2 = """
In data science, models are essential. The bigram language model is a simple model. 
It helps in understanding word pair frequencies. Data science projects can involve large text corpora.
"""

model2 = build_bigram_model(corpus2)
print("Generating text with model2 starting with 'data':")
print(generate_text(model2, "data", 15))

print("\nPrediction of next words for 'model':")
predict_next_words(model2, "model")
```
